```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

colori = sns.color_palette("pastel")

#vorremmo leggere tutta la tabella
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)

df = pd.read_csv("LabDelSole.csv", encoding = "latin1")
print("DataFrame")
df.head()
```

```python
print("INFO")
df.info()
print("\nDESCRIBE")
df.describe()
```

```python
df["tipologia"].value_counts()
```

```python
np.nanmean(df["peso"])
```

```python
# qui voglio riempire tutti nan con zeri tranne la colonna del peso (peso 0 è diverso da non pesato)
salvapeso = df["peso"]
df = df.fillna(0)
df["peso"] = salvapeso
```

```python
#qui invece vogliamo renderlo multi-index
nuove_col = []

for col in df.columns:
    if "passaparola" in col:
        nuove_col.append(("passaparola", col.replace("passaparola ", "")))
    elif "mercatino" in col:
        nuove_col.append(("mercatino", col.replace("mercatino ", "")))
    else:
        nuove_col.append(("info", col))

df.columns = pd.MultiIndex.from_tuples(nuove_col)
```

```python
df_venduti=df["passaparola"].add(df["mercatino"], fill_value=0)
df_venduti.columns = pd.MultiIndex.from_product([["venduti"], df_venduti.columns])
df = pd.concat([df, df_venduti], axis=1)
df.head()
```

```python
prezzo_tot = df_info["prezzo"].sum()
ore_tot = df_info["ore impiegate"].sum()
prezzo_ora_medio = prezzo_tot / ore_tot
prezzo_ora_medio
```

```python
df_info.head()
```

```python
df_milk = pd.DataFrame()
df_milk["tipologia"] = df_info["tipologia"]
df_milk["ore"] = df_info["ore impiegate"]
df_milk["prezzo"] = df_info["prezzo"]
df_milk["quantita"] = df["passaparola"].sum(axis=1) + df["mercatino"].sum(axis=1) + df_info["rimaste"] + df_info["regalate"]
df_milk["totale ore"] = df_milk["ore"] * df_milk["quantita"]
df_milk["totale guadagno"] = df_milk["prezzo"] * df_milk["quantita"]
df_milk["prezzo ora"] = df_milk["totale guadagno"] / df_milk["totale ore"]
df_milk["conviene?"] = df_milk["prezzo ora"] > prezzo_ora_medio
df_milk.head()
```

```python
df_quantita = df_milk.groupby(["prezzo", "ore"], as_index=False).agg({
    "quantita": "sum",
    "conviene?": "first" #tanto sono tutti uguali
})
df_quantita
```

```python
#giuro che è l'ultimo grafico
fig, ax = plt.subplots(figsize=(9, 6))

mask = df_quantita["conviene?"]

ax.scatter(df_quantita.loc[mask, "ore"], 
           df_quantita.loc[mask, "prezzo"],
           s=df_quantita.loc[mask, "quantita"] * 30,
           color=colori[2], 
           alpha=0.7, 
           label="Conviene")

ax.scatter(df_quantita.loc[~mask, "ore"], 
           df_quantita.loc[~mask, "prezzo"],
           s=df_quantita.loc[~mask, "quantita"] * 30,
           color=colori[3], 
           alpha=0.7, 
           label="Non conviene")

ax.set_ylabel("Prezzo (€)")
ax.set_xlabel("Ore impiegate")
ax.set_title("Convenienza rispetto media prezzo-ora")
ax.legend()

plt.tight_layout()
plt.show()
```
