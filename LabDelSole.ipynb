```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#vorremmo leggere tutta la tabella
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)

df = pd.read_csv("LabDelSole.csv", encoding = "latin1")
print("DataFrame")
df.head()
```

```python
print("INFO")
df.info()
print("\nDESCRIBE")
df.describe()
```

```python
# qui voglio riempire tutti nan con zeri tranne la colonna del peso (peso 0 è diverso da non pesato)
salvapeso = df["peso"]
df = df.fillna(0)
df["peso"] = salvapeso

#df.drop(columns=["modello"], inplace=True)
```

```python
nuove_col = []

for col in df.columns:
    if "passaparola" in col:
        nuove_col.append(("passaparola", col.replace("passaparola ", "")))
    elif "mercatino" in col:
        nuove_col.append(("mercatino", col.replace("mercatino ", "")))
    else:
        nuove_col.append(("info", col))

df.columns = pd.MultiIndex.from_tuples(nuove_col)
```

```python
df_venduti=df["passaparola"].add(df["mercatino"], fill_value=0)
df_venduti.columns = pd.MultiIndex.from_product([["venduti"], df_venduti.columns])
df = pd.concat([df, df_venduti], axis=1)
df.head()
```
