```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_pickle("LabDelSole.pickle")

colori = sns.color_palette("pastel")
```

```python
df_t_tot=pd.DataFrame()
df_t_tot["tipologia"]=df["info", "tipologia"]
df_t_tot["mercatino"]=df["mercatino"].sum(axis=1)
df_t_tot["passaparola"]=df["passaparola"].sum(axis=1)
df_t_tot["regalate"]=df["info", "regalate"]
df_t_tot["rimaste"]=df["info", "rimaste"]

df_t_tot.head()
```

```python
df_tipi_totali=df_t_tot.groupby("tipologia").sum().T
df_tipi_totali
```

```python
#GRAFICO 4
#metterlo con sns dovrei fare un melt
ordine_tipi = ["ciondolo", "bracciale", "anello", "orecchini", "altro"]

ax = df_tipi_totali[ordine_tipi].plot(kind="bar", figsize=(12, 6), width=0.8, color=colori)

ax.set_title("Vendite per Tipologia", pad = 20, fontsize = 22, fontweight = "bold")
ax.set_xlabel("Destinazione", fontsize = 18, fontweight = "normal", labelpad = 15)
ax.set_ylabel("Pezzi venduti", fontsize = 18, fontweight = "normal", labelpad = 15)

ax.set_facecolor("#FFFFFF")
ax.grid(True, color = "#BBBBBB", linewidth = 0.5, linestyle="--")
ax.grid(True, axis = 'y', which = 'minor',linewidth = 0.3, linestyle="--")

ax.yaxis.set_major_locator(plt.MultipleLocator(5))
ax.yaxis.set_minor_locator(plt.MultipleLocator(2.5))
ax.set_ylim(0, 35)

ax.legend(title="Tipologia", loc="upper left")
plt.xticks(rotation=0)

plt.savefig("Immagini/Tipologie.svg", bbox_inches = "tight", pad_inches = 0.3, transparent = False)
plt.savefig("Immagini/Tipologie.png", bbox_inches = "tight", pad_inches = 0.3, transparent = False)
```
