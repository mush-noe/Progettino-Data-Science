```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_pickle("LabDelSole.pickle")

colori = sns.color_palette("pastel")

df_info=df["info"]
```

```python
#GRAFICO 3
#grafico della distribuzione dei pesi dei vari oggetti
fig, (ax1, ax2) = plt.subplots(nrows = 1, ncols = 2, figsize = (30, 10))
sns.set(style = "whitegrid", font_scale = 1.5)

ordine = ["ciondolo", "bracciale", "orecchini", "anello"]
#ATTENZIONE STIAMO ANDANDO A TRASCURARE LA CATEGORIA ALTRO

# grafichino 1

sns.boxplot(
    ax = ax1,
    data=df_info,
    x="tipologia",
    y="peso",
    order=ordine,
    hue="perline",
    palette={True: colori[2], False: colori[3]}
)

ax1.set_facecolor("#FFFFFF")

ax1.grid(True, color = "#BBBBBB", linewidth = 0.5, linestyle="--")
ax1.grid(True, axis = 'y', which = 'minor',linewidth = 0.3, linestyle="--")

ax1.set_title("Ci sono perline?", pad = 20, fontsize = 30, fontweight = "bold")
ax1.set_xlabel("Tipologia di gioiello", fontsize = 26, fontweight = "normal", labelpad = 15)
ax1.set_ylabel("Peso (g)", fontsize = 26, fontweight = "normal", labelpad = 15)

ax1.legend(title="Ci sono perline?", loc="upper right")

ax1.yaxis.set_major_locator(plt.MultipleLocator(2.5))
ax1.yaxis.set_minor_locator(plt.MultipleLocator(0.5))
ax1.set_ylim(-1, 19)

# grafichino 2

sns.boxplot(
    ax = ax2,
    data=df_info,
    x="tipologia",
    y="peso",
    order=ordine,
    hue="design mio",
    palette={True: colori[0], False: colori[1]}
)
ax2.set_facecolor("#FFFFFF")

ax2.grid(True, color = "#BBBBBB", linewidth = 0.5, linestyle="--")
ax2.grid(True, axis = 'y', which = 'minor',linewidth = 0.3, linestyle="--")

ax2.set_title("L'ho disegnato io?", pad = 20, fontsize = 30, fontweight = "bold")
ax2.set_xlabel("Tipologia di gioiello", fontsize = 26, fontweight = "normal", labelpad = 15)
ax2.set_ylabel("Peso (g)", fontsize = 26, fontweight = "normal", labelpad = 15)

ax2.yaxis.set_major_locator(plt.MultipleLocator(2.5))
ax2.yaxis.set_minor_locator(plt.MultipleLocator(0.5))
ax2.set_ylim(-1, 19)
ax2.legend(title="Design mio?", loc="upper right")

plt.savefig("Immagini/Peso.svg", bbox_inches = "tight", pad_inches = 0.3, transparent = False)
plt.savefig("Immagini/Peso.png", bbox_inches = "tight", pad_inches = 0.3, transparent = False)
```
